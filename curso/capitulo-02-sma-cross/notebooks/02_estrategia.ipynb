# 02 - Estrategia: Cruce de Medias Móviles (SMA Cross)

**Capítulo**: 02 - SMA Cross

**Objetivo**: Implementar la estrategia de cruce de medias móviles con señales long/short.

---

```python
import sys
sys.path.insert(0, '../..')

import pandas as pd
import matplotlib.pyplot as plt
from curso.lib.data import download_historical
from curso.lib.indicators import sma

plt.style.use('seaborn-v0_8-whitegrid')
print('Setup completado ✓')
```

## 1. Datos

```python
TICKER = 'SPY'
df = download_historical(TICKER)
print(f'{TICKER}: {len(df)} registros')
```

## 2. Cálculo de doble SMA

```python
N1, N2 = 10, 20  # Defaults de spec.md

df['SMA_fast'] = sma(df, length=N1)
df['SMA_slow'] = sma(df, length=N2)

# Detectar cruces
df['golden_cross'] = (df['SMA_fast'] > df['SMA_slow']) & (df['SMA_fast'].shift(1) <= df['SMA_slow'].shift(1))
df['death_cross'] = (df['SMA_fast'] < df['SMA_slow']) & (df['SMA_fast'].shift(1) >= df['SMA_slow'].shift(1))

print(f'Golden crosses: {df["golden_cross"].sum()}')
print(f'Death crosses: {df["death_cross"].sum()}')
```

## 3. Visualización

```python
data_plot = df.tail(500)
fig, ax = plt.subplots(figsize=(14, 7))
ax.plot(data_plot.index, data_plot['Close'], linewidth=1, color='#333', label='Precio')
ax.plot(data_plot.index, data_plot['SMA_fast'], linewidth=1.2, color='#4CAF50', label=f'SMA({N1})')
ax.plot(data_plot.index, data_plot['SMA_slow'], linewidth=1.2, color='#F44336', label=f'SMA({N2})')

gc = data_plot[data_plot['golden_cross']]
dc = data_plot[data_plot['death_cross']]
ax.scatter(gc.index, gc['Close'], marker='^', color='green', s=100, zorder=5, label='Golden Cross')
ax.scatter(dc.index, dc['Close'], marker='v', color='red', s=100, zorder=5, label='Death Cross')

ax.set_title(f'{TICKER} - SMA Cross ({N1}/{N2})', fontsize=14)
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
