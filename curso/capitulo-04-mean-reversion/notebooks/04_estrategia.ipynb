# 04 - Estrategia: Reversión a la Media con Bollinger Bands

**Capítulo**: 04 - Mean Reversion

**Objetivo**: Implementar compra en banda inferior + salida en banda media, con stop ATR dinámico.

---

```python
import sys
sys.path.insert(0, '../..')

import pandas as pd
import matplotlib.pyplot as plt
from curso.lib.data import download_historical
from curso.lib.indicators import bollinger_bands, atr, sma

plt.style.use('seaborn-v0_8-whitegrid')
print('Setup completado ✓')
```

## 1. Datos

```python
TICKER = 'XLF'  # Sector financiero - buen comportamiento range-bound
df = download_historical(TICKER)
print(f'{TICKER}: {len(df)} registros')
```

## 2. Calcular Bollinger Bands + ATR

```python
BB_PERIOD = 20
BB_STD = 2.0
ATR_PERIOD = 14

bb = bollinger_bands(df, length=BB_PERIOD, std=BB_STD)
df['BB_upper'] = bb['upper']
df['BB_middle'] = bb['middle']
df['BB_lower'] = bb['lower']
df['ATR'] = atr(df, length=ATR_PERIOD)

# Señales
df['entry_signal'] = df['Close'] < df['BB_lower']
df['exit_signal'] = df['Close'] >= df['BB_middle']

print(f'Señales de entrada (toque banda inferior): {df["entry_signal"].sum()}')
print(f'ATR promedio: {df["ATR"].mean():.2f}')
```

## 3. Visualización con bandas

```python
data = df.tail(300).copy()

fig, ax = plt.subplots(figsize=(14, 7))
ax.plot(data.index, data['Close'], linewidth=1, color='#333', label='Precio')
ax.plot(data.index, data['BB_middle'], color='#2196F3', linestyle='--', linewidth=1, label='BB Media')
ax.plot(data.index, data['BB_upper'], color='#F44336', linestyle=':', linewidth=0.8, label='BB Superior')
ax.plot(data.index, data['BB_lower'], color='#4CAF50', linestyle=':', linewidth=0.8, label='BB Inferior')
ax.fill_between(data.index, data['BB_lower'], data['BB_upper'], alpha=0.05, color='blue')

entries = data[data['entry_signal']]
ax.scatter(entries.index, entries['Close'], marker='^', color='green', s=100, zorder=5, label='Entrada')

ax.set_title(f'{TICKER} - Bollinger Bands ({BB_PERIOD}, {BB_STD}σ)', fontsize=14)
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
